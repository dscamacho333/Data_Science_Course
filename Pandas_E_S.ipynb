# CSV

## Reading CSV (Importing)

```python
# Library to handle different file types and handle data

import pandas as pd
```

```python
# Library to check what encoding file is under

import chardet
```

```python
# Find out what encoding file is under (Important to avoid encoding exceptions)

with open('/content/drive/MyDrive/Programación Profesional/Data Science/ETL/Pandas (ETL)/Pandas E S/Datasets/CSV Datasets/superstore_data.csv', 'rb') as file:

  print(chardet.detect(file.read()))
```

```python
# How to import CSV separated by colon

super_store_data_colon_separated : pd.DataFrame = pd.read_csv(
    filepath_or_buffer='/content/drive/MyDrive/Programación Profesional/Data Science/ETL/Pandas (ETL)/Pandas E S/Datasets/CSV Datasets/superstore_data.csv', # Param to indicate what file we will be reading
    sep=',', # Param to indicate what the separator for the csv is
    usecols=['Id', 'Year_Birth', 'Education'], # Param to select chosen columns (could be the name as string or position as int)
    nrows=5, # Param to import n rows only,
    encoding='ascii' # Param to inidcate what encoding file has
)
```

```python
super_store_data_colon_separated
```

```python
# How to import CSV separated by semicolon

super_store_data_semicolon_separated : pd.DataFrame = pd.read_csv(
    '/content/drive/MyDrive/Programación Profesional/Data Science/ETL/Pandas (ETL)/Pandas E S/Datasets/CSV Datasets/superstore_data_punto_coma.csv',
    sep=';'
)
```
