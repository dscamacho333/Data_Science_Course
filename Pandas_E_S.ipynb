# SQL

## Creating SQL Connection

```python
# Import SQLAlchemy ORM

import sqlalchemy
```

```python
# Get specific libraries

from sqlalchemy import create_engine # To create Datasource connection
from sqlalchemy import MetaData # To get database metadata
from sqlalchemy import Table # To be able to handle SQL tables
from sqlalchemy import inspect # To inspect database
from sqlalchemy import text # To be able to do native SQL queries
```

```python
# Install MySQL driver package

!pip install pymysql

"""
DRIVERS FOR EACH SQL DATABASE
SQLite: It is integrated with Colab
MySQL: pymysql
Oracle: oracledb
Postgres: psycopg2-binary
SQL Server: pyodbc
"""
```

```python
# Define Database URL, it will always have the following format:
# (databaseName+driverName://user:password@host:port/dbName)
# Sometimes the URL Chnages depending on the DB, but generally follows this structure

DATABASE_CONNECTION_URL : str = 'sqlite:///:memory:'
```

```python
# Create database connection (MySQL)

engine : sqlalchemy.engine.base.Engine = create_engine(
    DATABASE_CONNECTION_URL,
    pool_pre_ping=True,
    echo=True
)
```

```python
engine
```

## Reading SQL (Importing)

```python
employee_rent_category_query : str = (
"""
SELECT
  *
FROM
  bank_customers bc
WHERE
  bc.Categoria_de_renta = 'Empleado'
"""
)
```

```python
employee_rent_category_query_result_df : pd.DataFrame = pd.read_sql(
    sql=text(employee_rent_category_query),
    con=engine.connect()
)
```

```python
employee_rent_category_query_result_df
```

```python
# Create new table

employee_rent_category_query_result_df.to_sql(
    'employees',
    index=False,
    con=engine.connect()
)
```

```python
# Read Table

pd.read_sql_table(
    'employees',
    con=engine.connect()
)
```

```python
pd.read_sql(
    sql=text(all_bank_customers_query),
    con=engine.connect()
)
```

```python
# Importing library to be able identify errors

from sqlalchemy.exc import SQLAlchemyError
```

```python
deletion_query : str = (
    """
    DELETE
    FROM
      bank_customers
    WHERE
      ID_Cliente = 5008804
    """
)
```

```python
# Delete record

try:

  pd.read_sql_query(
      sql=deletion_query,
      con=engine.connect()
  )

except SQLAlchemyError as e:

  print(e)
```

```python
# Confirm deletion

pd.read_sql(
    sql=text("SELECT * FROM bank_customers bc WHERE bc.ID_Cliente = 5008804 "),
    con=engine.connect()
)
```

```python
# Update query

update_query : str = (
    """
    UPDATE
      bank_customers
    SET
      Grado_estudio = 'Nivel superior'
    WHERE
      ID_Cliente = 5008808
    """
)
```

```python
# Update record

try:

  pd.read_sql_query(
      sql=update_query,
      con=engine.connect()
  )

except SQLAlchemyError as e:

  print(e)
```

```python
# Confirm update

pd.read_sql(
    sql=text("SELECT * FROM bank_customers bc WHERE bc.ID_Cliente = 5008808 "),
    con=engine.connect()
)
```
