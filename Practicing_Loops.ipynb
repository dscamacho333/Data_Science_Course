01 - Mostrar mensaje n veces

```python
# Define function

def show_message(
    number_of_iterations : int
):

  for i in range(0, number_of_iterations, 1):

    print(
        'Welcome to Buscante'
    )
```

```python
# Use function

number_of_iterations : int = int(
    input('Input number of iterations you want: ')
)

show_message(
    number_of_iterations
)
```

02- Calcular suma de números

```python
# Define function

def sum_numbers(
    numbers : list[int]
) -> int:

  total : int = 0

  for number in numbers:

    total = (total + number)

  return total
```

```python
# use function

numbers : list[int] = [
    1,
    2,
    3,
    4,
    5,
    6
]

total : int = sum_numbers(
    numbers
)

print(
    f'Total is: {total}'
)
```

03 - Organizando portafolio

```python
# Define fuction

def sort_portafolio(
    projects : list[str]
):

  for project in projects:

    if (project is not None):

      print(project)

    else:

      print('Missing project')
```

```python
# Use function

projects = ["web page", "game", "data analysis", None, "mobile app"]

sort_portafolio(
    projects
)
```

04 - Buscando libros

```python
# Define book class

class Book:

  author : str
  title : str
  qty : int

  def __init__(self, author, title):

    self.author = author
    self.title = title

  def __str__(self) -> str:
    return f'Book´s author is: {self.author} \nBook´s title is: {self.title}'


# Define function

def find_book_by_name(
    book_title : str,
    books : list[Book]
) -> Book:

  for book in books:

    if (book.title == book_title):

      return book


  return None
```

```python
# Use function

book_title : str = 'One Hundred Years of Solitude'

books = [
    Book("George Orwell", "1984"),
    Book("Gabriel García Márquez", "One Hundred Years of Solitude"),
    Book("Antoine de Saint-Exupéry", "The Little Prince"),
    Book("J.R.R. Tolkien", "The Hobbit"),
    Book("Jane Austen", "Pride and Prejudice")
]

book : Book = find_book_by_name(
    book_title,
    books
)

print(
    book
)
```

05 - Control de inventario

```python
# Define function

def simulate_sell(
    book : Book
):

  while ( book.qty != 0 ):

    print('Sold successfully')

    book.qty = ( book.qty - 1 )

    print(f'New book qty is: {book.qty}')
```

```python
# Use function

book : Book = Book(
    'JK Rowling',
    'Harry Potter'
)

book.qty = 10

simulate_sell(
    book
)
```
