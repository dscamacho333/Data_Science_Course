El interpretador de Python contiene una serie de funciones incorporadas que pueden ser invocadas es cualquier momento. Algunas que vamos a utilizar a lo largo de este curso son: `type()`, `print()`, `list()`, `zip()`, `sum()`, `map()`, etc.

#### **Situación 1:**

La institución educativa para la cual nos encontramos trabajando compartió los datos de las notas de un estudiante para que pudiesemos calcular su promedio con una casilla decimal.

Los datos recibidos corresponden a un diccionario cuyas llaves indican el trimestre en cuestión y sus valores corresponden a las notas de cada trimestre del estudiante en una respectiva materia.

````python
def calculate_average(
    scores : dict[str, str]
) -> float:

  sum_of_scores : float = sum(
      scores.values()
  )

  average = sum_of_scores / len(scores)

  return average
````

````python
scores : dict[str, float] = {
    '1T' : 9.5,
    '2T' : 8.0,
    '3T' : 7.0
}


average : float = calculate_average(
    scores
)

message : str = f'Avergae score is : {average}'

print(
    message
)
````

*Redondear el promedio usando round():*

https://docs.python.org/es/3/library/functions.html#round

````python
rounded_average : float = round(
    average,
    1
)

message : str = f'Avergae score is : {rounded_average}'

print(
    message
)
````

#### **Situación 6:**

Recibimos la demanda de transformar una lista con el nombre y las notas de tres trimestres de algunos estudiantes en una lista simple, con los nombres separados de las notas, y una lista de listas con las tres notas de cada estudiante separadas entre sí. Los datos recibidos corresponden a una lista con los nombres y las respectivas notas de cada estudiante.

Para facilitar nuestra comprensión del proceso, vamos a trabajar con un grupo de 5 estudiantes; sin embargo, puedes probar otros casos para practicar.


````python
notas_grupo = ['Juan', 8.0, 9.0, 10.0, 'Maria', 9.0, 7.0, 6.0, 'José', 3.4, 7.0, 7.0, 'Claudia', 5.5, 6.6, 8.0, 'Ana', 6.0, 10.0, 9.5]
````

````python
# Old version
separated_scores : list[float] = []

for i in range(0, len(scores), 1):

  if (i == 0 or i % 3 == 0):

    student_scores : list = []
    separated_scores.append(
        student_scores
    )
    student_scores.append(
        scores[i]
    )

  else:

    student_scores.append(
        scores[i]
    )

print(
    separated_scores
)
````

````python
# Optimized version

separated_scores : list[float] = []

for i in range(0, len(scores), 1):

  if (i == 0 or i % 3 == 0):

    student_scores : list = []
    separated_scores.append(
        student_scores
    )


  student_scores.append(
      scores[i]
  )



print(
    separated_scores
)
````

#### **Situación 7:**

Necesitamos generar una lista de tuplas con los nombres de los estudiantes y el código ID de cada uno de ellos para la plataforma de análisis de datos. La creación del código consiste en concatenar la primera letra del nombre del estudiante con un número aleatorio de 0 a 999. Los datos recibidos corresponden a una lista con los nombres de cada estudiante.

**¿Vamos a resolver este desafío?**

Para facilitar nuestra comprensión del proceso, vamos a trabajar con un grupo de 5 estudiantes; sin embargo, puedes probar otros casos para practicar.


````python
def generate_id(
    name : str,
    number : int
) -> str:

  first_name_letter : str  = name[0]

  number_as_string : str = str(
      number
  )

  return first_name_letter + number_as_string
````

````python
student_codes : list[tuple] = []
````

````python
from random import randint


for name in names:

  student_id = generate_id(
      name,
      randint(
          0,
          999
      )
  )

  student_codes.append(
      (name, student_id)
  )
````

````python
print(
    student_codes
)
````

Es una forma simple y concisa de crear una lista. Podemos aplicar condicionales y lazos para crear diversos tipos de listas a partir de patrones que deseamos para nuestra estructura de datos.

#### **Situación 8:**

Recibimos la demanda de crear una lista con el promedio de los estudiantes de la lista de listas que creamos en la Situación 6, redondeando el promedio a una casilla decimal. Recordando que cada lista de la lista de listas contiene las tres notas de cada estudiante.

**¿Vamos a resolver este desafío?**

**Tip:** Utiliza el formato:
```python
[expresion for item in lista]
```

````python
def promedio(lista: list=[0]) -> float:
  ''' Función para calcular el promedio de notas en una lista

  lista: list, default [0]
    Lista con las notas para calcular el promedio
  return = calculo: float
    Promedio calculado
  '''

  calculo = sum(lista) / len(lista)

  return calculo
````

````python
separated_scores
````

````python
averages : list[float] = [
    round(
        promedio(
            score
        ),
        1
    )

    for score in separated_scores
]

print(
    averages
)
````

````python
# Define Person Class

class Person:

  id : int = 1
  first_name : str
  lastname : str
  age : int

  def __init__(
      self,
      first_name,
      lastname,
      age
  ):
    self.id = Person.id
    self.first_name = first_name
    self.lastname = lastname
    self.age = age

    Person.id += 1

  def __str__(self) -> str:
    return (
      f"""
      Person ID: {self.id} \n
      Person name: {self.first_name} \n
      Person lastname: {self.lastname}
      """
    )

  def __eq__(self, other) -> bool:
    return isinstance(other, Person) and self.id == other.id
````

````python
# Define in memory Empty Datasource

people : list[ Person ] = []
````

````python
# Simple in memory CRUD

# Create Person in Memory
def create(
    person : Person
) -> Person:

  people.append(
      person
  )

  return person

# Read Person in Memory
def read(
    person_id : int
) -> Person:

  for person in people:

    if ( person.id == person_id ):

      return person


  raise Exception(f'Person with ID: {person_id} does not exist!')

# Update Person in Memory
def update(
    person_id : int,
    first_name : str,
    lastname : str,
    age : int
) -> Person:


  person_to_update : Person = read(
      person_id
  )

  person_to_update .id = person_id
  person_to_update.first_name = first_name
  person_to_update.lastname = lastname
  person_to_update.age = age

  delete(
      person_id
  )

  create(
      person_to_update
  )

  return person_to_update

# Delete Person in Memory
def delete(
    person_id
) -> None:

  person_to_delete : Person = read(
      person_id
  )

  people.remove(
      person_to_delete
  )

# Read All People in Memory
def read_all() -> list[Person]:
  return people
````

````python
# Function to handle create action
def handle_create():

  first_name : str = input('Firstname: ')
  lastname : str = input('Lastname: ')
  age : int = int(
      input('Age: ')
  )

  person : Person  = Person(
      first_name,
      lastname,
      age
  )

  person = create(
      person
  )

  print(
      person
  )



# Function to handle read action
def handle_read():

  person_id : int = int(
      input('ID: ')
  )

  try:

    person : Person = read(
        person_id
    )

    print(
        person
    )

  except Exception as e:

    print(
        e
    )



# Function to handle update action
def handle_update():

  id : int = int(
      input('Id: ')
  )

  first_name : str = input('Firstname: ')
  lastname : str = input('Lastname: ')
  age : int = int(
      input('Age: ')
  )


  try:

    person = update(
        id,
        first_name,
        lastname,
        age
    )

  except Exception as e:

    print(
        e
    )

  else:

    print(
        person
    )


# Function to handle delete action
def handle_delete():

  id : int = int(
      input('Id: ')
  )

  try:

    delete(
        id
    )

  except Exception as e:

    print(
        e
    )


# Function to handle read all action
def handle_read_all():

  people : list[Person] = read_all()

  for person in people:

    print(
        person
    )
````

````python
def check_what_to_handle(
    option : int
) -> None:

  match (option):

    case 1:

      handle_create()

    case 2:

      handle_read()

    case 3:

      handle_update()

    case 4:

      handle_delete()

    case 5:

      handle_read_all()
````

## Polimorfismo

````python
"""
Para el caso de Java existen dos tipos de polimorfimo:

1) Sobreescritura de métodos
2) Sobrecarga de me´todos

En python solo existe la sobreescritura de métodos

"""

# Clase padre con método
class AnimalDomestico():

  nombre : str

  def __init__(self, nombre):

    self.nombre = nombre

  def hacer_sonido(self) -> str:
      return 'Hago un sonido'

# Clase hija 01 sobrescribiendo método padre
class Perro(AnimalDomestico):

  def __init__(self, nombre):

    super().__init__(nombre)

  def hacer_sonido(self) -> str:
    return 'Guau!'

# Clase hija 02 sobrescribiendo método padre
class Gato(AnimalDomestico):

  def __init__(self, nombre):

    super().__init__(nombre)

  def hacer_sonido(self) -> str:
    return 'Miau!'
````

## Encapsulamiento

````python
class Persona:

  # Atributos de clase
  nombre : str
  apellido : str
  edad : int
  altura : float
  peso : float

  # Public (Dejarlos sin nada)
  # Protected ( _atributo )
  # Private ( __atributo )
  def __init__(self, nombre, apellido, edad, altura, peso):

    self.__nombre = nombre
    self.__apellido = apellido
    self.__edad = edad
    self.__altura = altura
    self.__peso = peso

  # Métodos de clase

  def hablar(self):

    print('Estoy hablando!')

  # Métodos Getters & Setters

  def get_nombre(self) -> str:
    return self.__nombre

  def set_nombre(self, nombre : str) -> None :
    self.__nombre = nombre

  def get_apellido(self) -> str:
    return self.__apellido

  def set_apellido(self, apellido : str) -> None:
    self.__apellido = apellido

  # Métodos Getters & Setters haciendo uso de decoradores (buena practica)

  @property
  def edad(self) -> int:
    return self.__edad

  @edad.setter
  def edad(self, edad : int) -> None:
    self.__edad = edad

  # Método toString()

  def __str__(self) -> str:
    return (
        f"""
        ----------------------------------
                  Datos Persona
        ----------------------------------
        Nombre: {self.__nombre}
        Apellido: {self.__apellido}
        Edad: {self.__edad}
        Altura: {self.__altura}
        Peso: {self.__peso}
        -----------------------------------
        """
    )
````

````python
miguel : Persona = Persona(
    'Miguel',
    'Alvarado',
    24,
    1.75,
    76
)

# Good coz it is toString() method
print(
    miguel
)

# Good coz it is private but has an accessor method
print(
    miguel.get_nombre()
)

miguel.set_nombre(
    'Miguel Angel'
)

print(miguel.get_nombre())

print(
    miguel.edad
)

# Error coz it is private
print(
    miguel.peso
)
````

## Abstracción

````python
# Importar ABC (indicar que clase es abstracta) y decorador abstractmethod (indicar método es abstracto)
from abc import ABC, abstractmethod

# Clase padre (abstracta)
class PersonaAbstracta(ABC):

  nombre : str
  apellido : str
  edad : int

  def __init__(self, nombre, apellido, edad):

    self.nombre = nombre,
    self.apellido = apellido
    self.edad = edad

  #Método normal
  def presentarse(self):

    print(
        f'Hola, mi nombre es {self.nombre} y tengo {self.edad}'
    )

  #Método abstracto
  @abstractmethod
  def realizar_actividad(self):
    pass

# Clase hija
class Estudiante(PersonaAbstracta):

  def __init__(self, nombre, apellido, edad):

    super().__init__(nombre, apellido, edad)

  def realizar_actividad(self):
    print('Ando estudiando')
````

````python
# Tratar de crear un objeto de PersonaAbstracta (Error porque las clases abstractas no se puede instanciar)

persona_abstracta : PersonaAbstracta = PersonaAbstracta(
    'Juan',
    'Rivera',
    32
)
````

````python
# Instanicar clase hija y probar métodos

juan_estudiante : Estudiante = Estudiante(
    'Juan',
    'Rivera',
    32
)

juan_estudiante.presentarse()

juan_estudiante.realizar_actividad()
````
