# Data Analysis

## Total Income

```python
store_one_income : float = float(
    store_one_dataframe.Precio.sum()
)
```

```python
store_two_income : float = float(
    store_two_dataframe.Precio.sum()
)
```

```python
store_three_income : float = float(
    store_three_dataframe.Precio.sum()
)
```

```python
store_four_income : float = float(
    store_four_dataframe.Precio.sum()
)
```

```python
print(f'Store 01: {store_one_income}')
print(f'Store 02: {store_two_income}')
print(f'Store 03: {store_three_income}')
print(f'Store 04: {store_four_income}')
```

```python
stores : list[str] = [
    'Store 01',
    'Store 02',
    'Store 03',
    'Store 04'
]

incomes : list[float] = [
    store_one_income,
    store_two_income,
    store_three_income,
    store_four_income
]
```

```python
fig, ax = plt.subplots()
bars = ax.bar(
    stores,
    incomes
)

ax.set_title('Ingreso total por tienda')
ax.set_xlabel('Tienda')
ax.set_ylabel('Ingreso total')

ax.ticklabel_format(style='plain', axis='y')
ax.bar_label(bars, fmt='%.0f')

plt.show()
```
