 01 - Monitoreo de ventas de comercio

```python
# Define function to compare sales.

def monitor_sales(
    apple_sold_qty : int,
    banana_sold_qty : int
) -> str:

  message : str = None

  if ( apple_sold_qty > banana_sold_qty ):

    message = 'Apples sold more'

  elif ( apple_sold_qty < banana_sold_qty ):

    message = 'Bananas sold more'

  else:

    message = 'There is a tie'

  return message
```

```python
# Use created function

apple_sold_qty : int = int(
    input('Input sold apples quantity: ')
)

banana_sold_qty : int = int(
    input('Input sold bananas quantity: ')
)

message : str = monitor_sales(
    apple_sold_qty,
    banana_sold_qty
)

print(
    message
)
```

02 - Temperatura de los servidores

```python
# Define function

def monitor_temperature(
    limit_temp : float,
    current_temp : float
) -> str:

  message : str = None

  if ( limit_temp < current_temp ):

    message = 'Alert'

  else:

    message = 'Everything ok'

  return message
```

```python
# Use function

limit_temp : float = float(
    input('Input limit temp: ')
)

current_temp : float = float(
    input('Input current temp: ')
)

message : str = monitor_temperature(
    limit_temp,
    current_temp
)

print(
   message
)
```

03 - Calculando tiempo total del proyecto

```python
# Define function

def calculate_total_proyect_time() -> str:

  message : str = None
  total_hours : int = 0

  for i in range(0,3,1):

    hours : int = int(
        input('Input project hours: ')
    )

    if ( hours <= 0 ):

      message = 'Hours can neither be zero nor negative'
      return message

    else:

      total_hours = (total_hours + hours)

  message = f'Total project hours: {total_hours}'

  return message
```

```python

# Use function

message : str = calculate_total_proyect_time()

print(
  message
)
```

04 - Calculando el IMC

```python
# Define first function

def calculate_imc(
    weight : float,
    height : float
) -> float :

  return ((weight)/(height**2))
```

```python
# Define second function

def determine_imc_situation(
    imc : float
) -> str:

  message : str = None

  if ( imc < 18.5 ):

    message = 'Under'

  elif ( 18.5 <= imc < 25  ):

     message = 'Normal'

  elif ( imc > 25 ):

     message = 'Above'

  return message
```

```python
# Use functions

weight : float = float(
     input('Input ur weight in kg: ')
)

height : float = float(
     input('Input ur height in m: ')
)

imc : float = calculate_imc(
     weight,
     height
)

print(
     f'Your IMC is: {imc}'
)

message : str = determine_imc_situation(
     imc
)

print(
     message
)
```

05 - Verificando acceso a beneficios sociales

```python
# Define function

def validate_access_to_social_benefits(
    income : float,
    number_of_kids : int
) -> str:

  message : str = None

  if ( income <= 2000 and number_of_kids > 0 ):

    message = 'Eligible'

  else:

    message = 'Inelegible'

  return message
```

```python
# Use function

income : float = float(
     input('Input ur income: ')
)

number_of_kids : int = int(
     input('Input # of kids: ')
)

message : str = validate_access_to_social_benefits(
     income,
     number_of_kids
)

print(
     message
)
```

06 - Evaluación de desempeño de los empleados

```python
# Define function

def get_employee_performance(
    points : int,
    years_of_employment : int
) -> str:

  message : str = None

  if ( points >= 7 ):

    if ( years_of_employment > 5 ):

      message = 'Promoted'

    else:

      message = 'Keep working'

  else:

    message = 'Improve'

  return message
```

```python
# Use function

points : int = int(
    input('Input employee points: ')
)

years_of_employment : int = int(
    input('Input employee years of experience: ')
)

message : str = get_employee_performance(
     points,
     years_of_employment
)

print(
     message
)
```
