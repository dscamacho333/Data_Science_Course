 01 - Monitoreo de ventas de comercio

```python
# Define function to compare sales.

def monitor_sales(
    apple_sold_qty : int,
    banana_sold_qty : int
) -> str:

  message : str = None

  if ( apple_sold_qty > banana_sold_qty ):

    message = 'Apples sold more'

  elif ( apple_sold_qty < banana_sold_qty ):

    message = 'Bananas sold more'

  else:

    message = 'There is a tie'

  return message
```

```python
# Use created function

apple_sold_qty : int = int(
    input('Input sold apples quantity: ')
)

banana_sold_qty : int = int(
    input('Input sold bananas quantity: ')
)

message : str = monitor_sales(
    apple_sold_qty,
    banana_sold_qty
)

print(
    message
)
```

02 - Temperatura de los servidores

```python
# Define function

def monitor_temperature(
    limit_temp : float,
    current_temp : float
) -> str:

  message : str = None

  if ( limit_temp < current_temp ):

    message = 'Alert'

  else:

    message = 'Everything ok'

  return message
```

```python
# Use function

limit_temp : float = float(
    input('Input limit temp: ')
)

current_temp : float = float(
    input('Input current temp: ')
)

message : str = monitor_temperature(
    limit_temp,
    current_temp
)

print(
   message
)
```

03 - Calculando tiempo total del proyecto

```python
# Define function

def calculate_total_proyect_time() -> str:

  message : str = None
  total_hours : int = 0

  for i in range(0,3,1):

    hours : int = int(
        input('Input project hours: ')
    )

    if ( hours <= 0 ):

      message = 'Hours can neither be zero nor negative'
      return message

    else:

      total_hours = (total_hours + hours)

  message = f'Total project hours: {total_hours}'

  return message
```

```python

# Use function

message : str = calculate_total_proyect_time()

print(
  message
)
```

04 - Calculando el IMC

```python
# Define first function

def calculate_imc(
    weight : float,
    height : float
) -> float :

  return ((weight)/(height**2))
```

```python
# Define second function

def determine_imc_situation(
    imc : float
) -> str:

  message : str = None

  if ( imc < 18.5 ):

    message = 'Under'

  elif ( 18.5 <= imc < 25  ):

     message = 'Normal'

  elif ( imc > 25 ):

     message = 'Above'

  return message
```

```python
# Use functions

weight : float = float(
     input('Input ur weight in kg: ')
)

height : float = float(
     input('Input ur height in m: ')
)

imc : float = calculate_imc(
     weight,
     height
)

print(
     f'Your IMC is: {imc}'
)

message : str = determine_imc_situation(
     imc
)

print(
     message
)
```

```python
# Define function

def validate_access_to_social_benefits(
    income : float,
    number_of_kids : int
) -> str:

  message : str = None

  if ( income <= 2000 and number_of_kids > 0 ):

    message = 'Eligible'

  else:

    message = 'Inelegible'

  return message
```

06 - Evaluación de desempeño de los empleados

```python
# Define function

def get_employee_performance(
    points : int,
    years_of_employment : int
) -> str:

  message : str = None

  if ( points >= 7 ):

    if ( years_of_employment > 5 ):

      message = 'Promoted'

    else:

      message = 'Keep working'

  else:

    message = 'Improve'

  return message
```

07 - Juegos de múltiplos

```python
# Define function

def validate_magic_number(
    number : int
) -> str:

  message : str = None

  if ( number % 3 == 0 and number % 5 == 0 ):

    message = 'Magic Number'

  elif ( number % 3 == 0 ):

    message = 'Divisible by 3'

  elif ( number % 5 == 0 ):

    message = 'Divisible by 5'

  else:

    message = 'No magic'

  return message
```

```python

# Use function

number : int = int(
    input('Input a number: ')
)

message : str = validate_magic_number(
     number
)

print(
     message
)
```

08 - Sistema para becas escolares

```python
# Define function

def determine_intership(
    total_family_income : float,
    student_gpa : float,
    assistance_percentage : float
) -> str:

  message : str = None

  if ( total_family_income < 1500 and student_gpa > 8 and assistance_percentage >= 0.90 ):

    message = 'Full intership'

  elif ( total_family_income < 2500 and student_gpa > 7 and assistance_percentage >= 0.85 ):

    message = 'Half imtership'

  else:

     message = 'No intership'

  return message
```

```python
# Use function

total_family_income : float = float(
    input('Input family income: ')
)

student_gpa : float = float(
    input('Input student gpa: ')
)

assistance_percentage : float = float(
    input('Input assistance percentage: ')
)

message : str = determine_intership(
    total_family_income,
    student_gpa,
    assistance_percentage
)

print(
    message
)
```

09 - Sistema de tarifas transporte público

```python
# Define function

def determine_fee(
    passenger_age : int,
    distance : int
) -> float:

  fee : float = 0.0

  if ( passenger_age < 6 ):

    fee = 0.0

  elif ( 6 <= passenger_age <= 18 ):

    if ( distance <= 20 ):

      fee = 1.5

    elif ( distance > 20 ):

      fee = 2.5

  else:

      if ( distance <= 20 ):

        fee = 2.5

      elif ( distance > 20 ):

        fee = 4.0

  return fee
```

```python
# Use function

passenger_age : float = float(
    input('Input your age: ')
)

distance : float = float(
    input('Input the distance in KM: ')
)

fee : float = determine_fee(
    passenger_age,
    distance
)

message : str = f'Fee to pay is: {fee}'

print(
    message
)
```

10 - Evaluación desempeño empresarial

```python
# Define function

def evaluate_enterprise_performance(
    total_income : float,
    total_expenses : float,
    number_of_new_customers : int
) -> str:

  utility : float = (total_income - total_expenses)
  message : str = None

  if ( utility > 10000 and number_of_new_customers >= 50 ):
    message = 'Excelent'
  elif ( utility > 5000 and number_of_new_customers >= 20 ):
    message = 'Good'
  elif( utility > 0 ):
    message = 'Regular'
  else:
    message = 'Bad'

  return message
```

```python
# Use function

total_income : float = float(
    input('Income: ')
)
total_expenses : float = float(
    input('Expenses: ')
)
number_of_new_customers : int = int(
    input('New customers: ')
)

message : str = evaluate_enterprise_performance(
    total_income,
    total_expenses,
    number_of_new_customers
)

print(
    message
)
```
